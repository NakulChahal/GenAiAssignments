## Task 2: Catagories (set)

```python

# Parallel categories list
categories = [
    "Electronics",
    "Electronics",
    "Electronics",
    "Accessories",
    "Accessories",
    "Accessories",
    "Electronics",
    "Office"
]

# 1. Create a set of categories
categories_set = set(categories)

print("Unique categories:", categories_set)

# 2. Add a new category and show duplicates ignored
categories_set.add("Gaming")
categories_set.add("Electronics")   

print("Updated categories set:", categories_set)

# 3. Check whether a category exists
print("Is 'Office' in categories set-",
      "Office" in categories_set)


# Total number of unique categories
print("Total unique categories:",
      len(categories_set))
```

## Task 3: Product Pricing

```python


# Creating dictionary
price_dict = {
    "Laptop": 55000,
    "Mouse": 800,
    "Keyboard": 1500,
    "Monitor": 12000
}

# Adding new item
price_dict["Printer"] = 7000

# Updating value
price_dict["Mouse"] = 1000

# Deleting item with error handling
item_to_remove = "Keyboard"

if item_to_remove in price_dict:
    del price_dict[item_to_remove]
    print(f"{item_to_remove} removed successfully.")
else:
    print(f"{item_to_remove} not found.")

# Calculating average price
average_price = sum(price_dict.values()) / len(price_dict)

print("Updated Dictionary:", price_dict)
print("Average Price:", average_price)
```

## Task 4: Combined Operations

```python


# Product catalog
products = [
    {"name": "Laptop", "category": "Electronics"},
    {"name": "Mouse", "category": "Accessories"},
    {"name": "Keyboard", "category": "Accessories"},
    {"name": "Monitor", "category": "Electronics"},
    {"name": "Chair", "category": "Furniture"}
]

# Extract unique categories using set
unique_categories = set()

for product in products:
    unique_categories.add(product["category"])

print("Unique Categories:", unique_categories)

# Creating category mapping
category_map = {}

for category in unique_categories:
    category_map[category] = []

for product in products:
    category_map[product["category"]].append(product["name"])

print("Category Mapping:")
print(category_map)
```

```python
# Find category with maximum products using for loop
max_category = ""
max_count = 0

for category in category_map:
    count = len(category_map[category])

    if count > max_count:
        max_count = count
        max_category = category

print("\nCategory with maximum products:", max_category)
print("Products:")

for item in category_map[max_category]:
    print(item)
```
